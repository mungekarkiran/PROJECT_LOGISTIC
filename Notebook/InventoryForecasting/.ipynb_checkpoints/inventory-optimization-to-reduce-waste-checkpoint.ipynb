# Inventory Optimization to Reduce Waste in Retail Supply Chains

```python
!pip install -q datasets

import os, warnings, json, shutil
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
warnings.filterwarnings("ignore"); sns.set_theme(style="whitegrid")
%matplotlib inline

BASE="/kaggle/working/outputs" if os.path.isdir("/kaggle/working") else "outputs"
EDA,RESULT,METRIC=[os.path.join(BASE,x) for x in ("eda","results","metrics")]
for d in (EDA,RESULT,METRIC): os.makedirs(d,exist_ok=True)

def mape(y,yh):
    y,yh=np.asarray(y,float),np.asarray(yh,float); m=y>0
    return np.mean(np.abs((y[m]-yh[m])/y[m]))*100
def wape(y,yh):
    y,yh=np.asarray(y,float),np.asarray(yh,float)
    return np.sum(np.abs(y-yh))/np.sum(np.abs(y))*100
def smape(y,yh):
    y,yh=np.asarray(y,float),np.asarray(yh,float); d=np.abs(y)+np.abs(yh); m=d!=0
    return np.mean(2*np.abs(y[m]-yh[m])/d[m])*100
METRIC_NAMES=["MAE","RMSE","MAPE","WAPE","sMAPE","R2"]
def metrics(y,yh):
    return (mean_absolute_error(y,yh),np.sqrt(mean_squared_error(y,yh)),
            mape(y,yh),wape(y,yh),smape(y,yh),r2_score(y,yh))
```

## 1. Load FreshRetailNet-50K from Hugging Face

```python
from datasets import load_dataset

ds = load_dataset("Dingdong-Inc/FreshRetailNet-50K", split="train")
# drop the hourly sequence columns: hours_sale sums to the target (leakage) and both are heavy
ds = ds.remove_columns([c for c in ["hours_sale","hours_stock_status"] if c in ds.column_names])
df = ds.to_pandas()
del ds
print("Raw shape:", df.shape)
print("Columns:", df.columns.tolist())
df.head()
```

## 2. Sample stores + aggregate to store x category x day

The raw data is 4.5M rows. We sample a set of stores and aggregate products up to the
**first category level per store per day**, which keeps the data real but makes the target
larger and more stable (and keeps runtime sane on Kaggle).

```python
N_STORES = 150          # sample size; raise for more data, lower for speed
USE_STOCKOUT_FEATURE = False   # realised outcome -> kept out of the predictors by default

df["dt"] = pd.to_datetime(df["dt"])
rng = np.random.default_rng(42)
stores = df["store_id"].unique()
keep = rng.choice(stores, size=min(N_STORES, len(stores)), replace=False)
df = df[df["store_id"].isin(keep)]
print("After store sampling:", df.shape)

grp = ["city_id","store_id","first_category_id","dt"]
data = df.groupby(grp).agg(
    sales=("sale_amount","sum"),
    discount=("discount","mean"),
    holiday_flag=("holiday_flag","max"),
    activity_flag=("activity_flag","max"),
    precpt=("precpt","mean"),
    avg_temperature=("avg_temperature","mean"),
    avg_humidity=("avg_humidity","mean"),
    avg_wind_level=("avg_wind_level","mean"),
    stockout_hours=("stock_hour6_22_cnt","mean"),
    n_products=("product_id","nunique"),
).reset_index()

TARGET="sales"
data["DayOfWeek"]=data["dt"].dt.dayofweek
data["Day"]=data["dt"].dt.day
data["Month"]=data["dt"].dt.month
periods=np.sort(data["dt"].unique())
tmap={d:i for i,d in enumerate(periods)}
data["t"]=data["dt"].map(tmap)
SPLIT_T=int(data["t"].max()*0.8)
train_mask=data["t"]<=SPLIT_T

feat=["discount","holiday_flag","activity_flag","precpt","avg_temperature","avg_humidity",
      "avg_wind_level","n_products","DayOfWeek","Day","Month",
      "city_id","store_id","first_category_id"]
if USE_STOCKOUT_FEATURE: feat.append("stockout_hours")

print(f"Aggregated shape: {data.shape} | days {data['t'].max()+1} | split at t={SPLIT_T}")
print("Train rows:",train_mask.sum()," Test rows:",(~train_mask).sum())
print("Features:",feat)
data[[TARGET]+feat].describe().T
```

## 3. Exploratory Data Analysis

```python
tot=data.groupby("t")[TARGET].sum()
plt.figure(figsize=(10,4)); plt.plot(tot.index,tot.values,color="#2b6cb0")
plt.axvline(SPLIT_T,color="red",ls="--",label="train/test split"); plt.legend()
plt.title("Daily Total Sales (sampled stores)"); plt.xlabel("Day index"); plt.ylabel("Sales")
plt.tight_layout(); plt.savefig(f"{EDA}/eda_daily_demand.pdf",dpi=130); plt.show()

plt.figure(figsize=(7,4))
data.groupby("first_category_id")[TARGET].mean().sort_values().tail(12).plot(kind="barh",color="#38a169")
plt.title("Average Daily Sales by Product Category (top 12)"); plt.xlabel("Avg Sales")
plt.tight_layout(); plt.savefig(f"{EDA}/eda_category.pdf",dpi=130); plt.show()

num=["discount","precpt","avg_temperature","avg_humidity","avg_wind_level",
     "stockout_hours","n_products","holiday_flag","activity_flag"]
plt.figure(figsize=(8,6))
sns.heatmap(data[num+[TARGET]].corr(),cmap="coolwarm",center=0)
plt.title("Correlation Heatmap"); plt.tight_layout()
plt.savefig(f"{EDA}/eda_corr_heatmap.pdf",dpi=130); plt.show()

plt.figure(figsize=(8,4))
data.groupby("DayOfWeek")[TARGET].mean().plot(kind="bar",color="#dd6b20")
plt.title("Average Sales by Day of Week (0=Mon)"); plt.ylabel("Avg Sales")
plt.tight_layout(); plt.savefig(f"{EDA}/eda_seasonality.pdf",dpi=130); plt.show()

# waste-relevant: how stockouts relate to sales
b=pd.cut(data["stockout_hours"],bins=[-.01,0.01,1,2,4,24],
         labels=["0","0-1","1-2","2-4","4+"])
plt.figure(figsize=(8,4))
data.groupby(b)[TARGET].mean().plot(kind="bar",color="#e53e3e")
plt.title("Average Sales vs Out-of-Stock Hours (censored demand)")
plt.xlabel("Mean out-of-stock hours (06:00-22:00)"); plt.ylabel("Avg Sales")
plt.tight_layout(); plt.savefig(f"{EDA}/eda_stockout.pdf",dpi=130); plt.show()
```

## 4. Tabular models (time-based split) — Linear Regression, Random Forest, XGBoost

```python
tr,te=data[train_mask],data[~train_mask]
Xtr,ytr=tr[feat].values,tr[TARGET].values
Xte,yte=te[feat].values,te[TARGET].values
sc=MinMaxScaler().fit(Xtr)
results={}
lr=LinearRegression().fit(sc.transform(Xtr),ytr)
results["Linear Regression"]=metrics(yte,lr.predict(sc.transform(Xte)))
rf=RandomForestRegressor(n_estimators=150,max_depth=20,n_jobs=-1,random_state=42).fit(Xtr,ytr)
rf_pred=rf.predict(Xte); results["Random Forest"]=metrics(yte,rf_pred)
from xgboost import XGBRegressor
xgb=XGBRegressor(n_estimators=400,max_depth=8,learning_rate=0.08,subsample=0.9,
                 colsample_bytree=0.9,n_jobs=-1,random_state=42).fit(Xtr,ytr)
results["XGBoost"]=metrics(yte,xgb.predict(Xte))
pd.DataFrame({k:results[k] for k in ["Linear Regression","Random Forest","XGBoost"]},
             index=METRIC_NAMES).T.round(3)
```

## 5. ARIMA — fitted per store-category series (same target, same split)

```python
from statsmodels.tsa.arima.model import ARIMA
a_pred,a_true=[],[]
for _,g in data.groupby(["store_id","first_category_id"]):
    g=g.sort_values("t")
    trv=g.loc[g.t<=SPLIT_T,TARGET].values; tev=g.loc[g.t>SPLIT_T,TARGET].values
    if len(trv)<20 or len(tev)==0: continue
    try: f=ARIMA(trv,order=(1,1,1)).fit().forecast(steps=len(tev))
    except Exception: f=np.repeat(trv.mean(),len(tev))
    a_pred.append(np.nan_to_num(np.asarray(f,float),nan=trv.mean())); a_true.append(tev)
results["ARIMA"]=metrics(np.concatenate(a_true),np.concatenate(a_pred))
print("ARIMA series fitted:",len(a_pred))
pd.DataFrame({"ARIMA":results["ARIMA"]},index=METRIC_NAMES).T.round(3)
```

## 6. LSTM — one global model across all store-category series

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,Dense
tf.random.set_seed(42)
W=14
Xs,ys,is_te=[],[],[]
for _,g in data.groupby(["store_id","first_category_id"]):
    g=g.sort_values("t"); v=g[TARGET].values.astype("float32"); ts=g["t"].values
    for i in range(len(v)-W):
        Xs.append(v[i:i+W]); ys.append(v[i+W]); is_te.append(ts[i+W]>SPLIT_T)
X_arr=np.array(Xs,dtype="float32"); y_arr=np.array(ys,dtype="float32"); m_te=np.array(is_te)
s2=MinMaxScaler().fit(X_arr[~m_te].reshape(-1,1))
Xs_s=s2.transform(X_arr.reshape(-1,1)).reshape(X_arr.shape)[...,None]
ys_s=s2.transform(y_arr.reshape(-1,1)).ravel()
model=Sequential([LSTM(64,input_shape=(W,1)),Dense(1)])
model.compile(optimizer="adam",loss="mse")
model.fit(Xs_s[~m_te],ys_s[~m_te],epochs=12,batch_size=128,verbose=0)
lstm_pred=s2.inverse_transform(model.predict(Xs_s[m_te],verbose=0)).ravel()
results["LSTM"]=metrics(y_arr[m_te],lstm_pred)
print("LSTM windows - train:",(~m_te).sum()," test:",m_te.sum())
pd.DataFrame({"LSTM":results["LSTM"]},index=METRIC_NAMES).T.round(3)
```

## 7. Final results + figures

```python
res_df=pd.DataFrame(results,index=METRIC_NAMES).T.round(3).sort_values("RMSE")
res_df.to_csv(f"{METRIC}/results_metrics.csv")
with open(f"{METRIC}/results_metrics.json","w") as f:
    json.dump({k:dict(zip(METRIC_NAMES,[float(x) for x in v])) for k,v in results.items()},f,indent=2)

res_df[["MAE","RMSE"]].plot(kind="bar",figsize=(8,4.5),color=["#4299e1","#2b6cb0"])
plt.title("Model Comparison: MAE and RMSE (lower is better)"); plt.ylabel("Error")
plt.xticks(rotation=20,ha="right"); plt.tight_layout()
plt.savefig(f"{RESULT}/results_error_bars.pdf",dpi=130); plt.show()

res_df["MAPE"].plot(kind="bar",figsize=(8,4),color="#e53e3e")
plt.title("Model Comparison: MAPE %"); plt.ylabel("MAPE (%)")
plt.xticks(rotation=20,ha="right"); plt.tight_layout()
plt.savefig(f"{RESULT}/results_mape.pdf",dpi=130); plt.show()

res_df[["WAPE","sMAPE"]].plot(kind="bar",figsize=(8,4),color=["#2f855a","#805ad5"])
plt.title("Model Comparison: WAPE and sMAPE % (robust metrics)"); plt.ylabel("Error (%)")
plt.xticks(rotation=20,ha="right"); plt.tight_layout()
plt.savefig(f"{RESULT}/results_wape_smape.pdf",dpi=130); plt.show()

res_df["R2"].sort_values().plot(kind="barh",figsize=(8,4),color="#2b6cb0")
plt.title("Model Comparison: $R^2$ (higher is better)"); plt.xlabel("$R^2$")
plt.tight_layout(); plt.savefig(f"{RESULT}/results_r2.pdf",dpi=130); plt.show()

pd.Series(rf.feature_importances_,index=feat).sort_values().tail(12).plot(
    kind="barh",figsize=(8,5),color="#805ad5")
plt.title("Random Forest Feature Importance (top 12)"); plt.tight_layout()
plt.savefig(f"{RESULT}/results_feat_importance.pdf",dpi=130); plt.show()

plt.figure(figsize=(6,6)); plt.scatter(yte,rf_pred,s=6,alpha=0.25,color="#2b6cb0")
lims=[0,max(yte.max(),rf_pred.max())]; plt.plot(lims,lims,"r--")
plt.xlabel("Actual"); plt.ylabel("Predicted"); plt.title("Random Forest: Actual vs Predicted")
plt.tight_layout(); plt.savefig(f"{RESULT}/results_actual_vs_pred.pdf",dpi=130); plt.show()

key=data.groupby(["store_id","first_category_id"])[TARGET].sum().idxmax()
g=data[(data.store_id==key[0])&(data.first_category_id==key[1])].sort_values("t")
gte=g[g.t>SPLIT_T]
plt.figure(figsize=(10,4))
plt.plot(g["t"],g[TARGET],label="Actual",color="black")
plt.plot(gte["t"],rf.predict(gte[feat].values),label="Random Forest",color="#2b6cb0")
plt.axvline(SPLIT_T,color="red",ls="--",label="split"); plt.legend()
plt.title(f"Daily Demand Forecast vs Actual - Store {key[0]}, Category {key[1]}")
plt.xlabel("Day index"); plt.ylabel("Sales"); plt.tight_layout()
plt.savefig(f"{RESULT}/results_ts_forecast.pdf",dpi=130); plt.show()

print("=== FINAL MODEL COMPARISON ==="); res_df
```
